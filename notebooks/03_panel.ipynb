# Building the panel

Sample rules: one treated and one control side, a first treatment year, no group already treated in 1992, and a control side that stays clean within five years of the treatment date. Then the within-pair light gap.

**Expected:** group 1179 (Waboni, KE/SO) drops as always treated, leaving 11 groups, 22 units and 726 rows. The pair panel has 363 rows. Control countries are Angola, Nigeria and Sudan.

```python
import sys
sys.path.append("../src")

import pandas as pd
import matplotlib.pyplot as plt

import config
import panel
```

```python
lights = pd.read_csv(config.WORK / 'ntl_by_unit_year.csv')
skeleton = pd.read_csv(config.WORK / 'unit_year_treatment.csv')
panels, pair = panel.run(lights, skeleton)
```

```python
panel.composition(pair)
```

```python
# check against the expected sample
df = panels['narrow']
assert df['G1ID'].nunique() == 11, df['G1ID'].nunique()
assert len(df) == 726, len(df)
assert len(pair) == 363, len(pair)
print('sample matches')
```
